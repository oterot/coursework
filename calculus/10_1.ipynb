```python
from matplotlib import pyplot as plt

def f(x):
  return 1/2**x

def plot_sequence(series, n):
  x = [i for i in range(1, n+1)]
  y = []

  for i in x:
    term = series(i)
    y.append(term)
    print(term)

  print()
  plt.scatter(x,y)

plot_sequence(f,10)
```

```python
def g(x):
  return ((-1)**x*x)/(x**2+1)

plot_sequence(g,10)
```

```python
def f(x):
  return 2 * x + 1

def implicit_plot(f,a,n):
  a = a
  print(a)
  x = [i for i in range(1,n+1)]
  y = [a]
  for i in range(1,n):
    term = f(a)
    y.append(term)
    print(term)
    a = term

  print()
  plt.scatter(x,y)

implicit_plot(f, 1, 10)
```

```python
def g(x):
  return 2 * x + 1

implicit_plot(g,-1,10)
```
