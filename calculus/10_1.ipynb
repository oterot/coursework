```python
from matplotlib import pyplot as plt

def f(x):
  return 1/2**x

def plot_explicit(f, n):
  x = [i for i in range(1, n+1)]
  y = []

  for i in x:
    term = f(i)
    y.append(term)
    print(term)

  print()
  plt.scatter(x,y)

plot_explicit(f,10)
```

```python
def g(x):
  return ((-1)**x*x)/(x**2+1)

plot_explicit(g,10)
```

```python
def f(x):
  return 2 * x + 1

def plot_implicit(f,a,n):
  print(a)
  x = [i for i in range(1,n+1)]
  y = [a]
  for i in range(1,n):
    term = f(a)
    y.append(term)
    print(term)
    a = term

  print()
  plt.scatter(x,y)

plot_implicit(f, 1, 10)
```

```python
def g(x):
  return 2 * x + 1

plot_implicit(g,-1,10)
```

```python
def f(x):
  return (4*x**3)/(x**3+1)

plot_explicit(f, 10)
```

The graph seems to suggest that this series converges, with the terms approaching 4.

```python
def f(x):
  return 1/(x**2 + x)

def partial_sum_plot(f, n):
  sum = 0
  x = [i for i in range(1, n+ 1)]
  y = []
  for i in x:
    sum += f(i)
    y.append(sum)
    print(sum)
  plt.scatter(x,y)

partial_sum_plot(f,10)

```

The graph suggests that the infinite series converges to 1.
